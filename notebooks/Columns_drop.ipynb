# Dropping columns- Stage 1

**Insight**: For dropping columns as there are 135 columns which are too many for model and training purpose. I'm reducing it to something nearly 80. For this purpose I divided columns as different grouping like some columns have described the same thing again and again some have more than 90% missing terms for e.g. attacktype2_txt (96.5% empty).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### ---- GROUP 1: -----------------------

```python
group1= [
    "attacktype2", "attacktype2_txt", "attacktype3", "attacktype3_txt",
    "targtype2", "targtype2_txt", "targsubtype2", "targsubtype2_txt",
    "corp2", "target2", "natlty2", "natlty2_txt",
    "targtype3", "targtype3_txt", "targsubtype3", "targsubtype3_txt",
    "corp3", "target3", "natlty3", "natlty3_txt",
    "weaptype2", "weaptype2_txt", "weapsubtype2", "weapsubtype2_txt",
    "weaptype3", "weaptype3_txt", "weapsubtype3", "weapsubtype3_txt",
    "weaptype4", "weaptype4_txt", "weapsubtype4", "weapsubtype4_txt",
    "gname2", "gsubname2", "gname3", "gsubname3",
    "guncertain2", "guncertain3",
    "claim2", "claimmode2", "claimmode2_txt",
    "claim3", "claimmode3", "claimmode3_txt",
]

assert len(group1) == 44
```

**Insight**: This group is focussed to the columns which is repeated to the same parent column like subparts but with no meaningfull insights from it also it have major portion of missing values.

### ---- GROUP 2: ------

```python
group2 = [
    "country", "region",
    "attacktype1", "targtype1", "targsubtype1", "natlty1",
    "weaptype1", "weapsubtype1",
]
assert len(group2) == 8
```

**Insight**: This group is focused to the group which has same information as the parent column but in numeric form. Hence, no meaning for keeping it.

```python
drop_cols = group1 + group2
assert len(set(drop_cols)) == 52
drop_cols = [c for c in drop_cols if c in df.columns]
```

```python
reduced = df.drop(columns=drop_cols)
print("Stage 1 shape:", reduced.shape)
```

```python
reduced.to_csv("../Data/globalterrorismdb_reduced_stage1.csv", index=False)
print("Saved: globalterrorismdb_reduced_stage1.csv")
```
